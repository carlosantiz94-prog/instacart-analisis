# Comprobación del Entorno
Este cuaderno verifica el correcto funcionamiento de Python y Pandas.

```python
import sys
import pandas as pd

print(sys.version_info.major, sys.version_info.minor)
print(pd.__version__)
```

```python
def franja(hora):
    if hora < 12:
        return 'mañana'
    elif hora < 19:
        return 'tarde'
    else:
        return 'noche'

print(f"8 -> {franja(8)}")
print(f"13 -> {franja(13)}")
print(f"21 -> {franja(21)}")
```
